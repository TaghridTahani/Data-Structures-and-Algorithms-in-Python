##The Brute Force Solution Would be :
####Cheacking every number one by one, and then when the query matches with a number in the list, and give that postion as an output. This would be the first thought of the solution , which also gives o(n)

```python
def locate_cards(cards, query):
  #create a variable postion with the value 0
  position = 0

  #set up a loop for repetition
  while True:

    #Check if the len is 0
    if len(cards) == 0:
      return -1

    #Check if element at th ecurrent position matche the query
    if cards[position] == query:

      #Answer found!, return and exist...
      return position

    #Increment the position
    position +=1

    #Check if we have reached the end of the array
    if position == len(cards):

      #Number not found , return -1
      return -1

```

#Leet Problem : 35. Search Insert Position
###Given a sorted array of distinct integers and a target value, return the index if the target is found. If not, return the index where it would be if it were inserted in order.

###Let's trace how your searchInsert code handles this exact example:
####*nums: [1, 3, 5, 6]
####*target: 7

###Step-by-Step Trace:
###1. Initial State:
####*left = 0, right = 3

###2.Iteration 1:
####*mid = (0 + 3) // 2 = 1
####*nums[1] is 3.
####*Is 3 == 7? No.
####*Is 3 < 7? Yes, so left = mid + 1 $\rightarrow$ left = 2, right = 3.

###3.Iteration 2:
####*left = 2, right = 3
####*mid = (2 + 3) // 2 = 5 // 2 = 2
####*nums[2] is 5
####*Is 5 == 7? No.
####*Is 5 < 7? Yes, so left = mid + 1 $\rightarrow$ left = 3, right = 3.


###4.Iteration 3:
####*left = 3, right = 3
####*mid = (3 + 3) // 2 = 3
####*nums[3] is 6
####*Is 6 == 7? No..
####*Is 6 < 7? Yes, so left = mid + 1 $\rightarrow$ left = 4, right = 3.

###5.End of Loop:
####*Now left = 4 and right = 3. The condition left <= right (4 <= 3) is False, so the loop terminates.
####*mid = (3 + 3) // 2 = 3
####*The function returns left, which is 4.


```python
class Solution:
    def searchInsert(self, nums: list[int], target: int) -> int:
        self.left = 0
        self.right = len(nums)-1
        mid = 0
        while self.left<=self.right:
            mid = (self.left+self.right)//2
            if nums[mid]==target:
                return mid
            elif nums[mid]<target:
                self.left= mid+1
            elif nums[mid]>target:
                self.right= mid-1

        return self.left

```
