##The Brute Force Solution Would be :
####Cheacking every number one by one, and then when the query matches with a number in the list, and give that postion as an output. This would be the first thought of the solution , which also gives o(n)

```python
def locate_cards(cards, query):
  #create a variable postion with the value 0
  position = 0
  #set up a loop for repetition
  while True:
    if cards[position] == query:
      return position
    position +=1
    if position == len(cards):
      return -1
```
